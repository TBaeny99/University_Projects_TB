```python
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd
import geopandas as gpd
from shapely.geometry import Polygon

# Set working directory
os.chdir(r'C:/Users/crist/OneDrive/Documents/UNFC/Predictive/Assignment2')

# Load the CSV file
df = pd.read_csv('Dataset_alzheimer.csv')

# Ensure the column name is correct
print(df.columns)  # Check available column names

# Data
if 'Age' in df.columns:  # Check if 'Age' exists in the DataFrame
    mean = np.mean(df['Age'])
    std_dev = np.std(df['Age'])

    # Detect outliers using Z-Score
    threshold = 3
    outliers_z_score = [x for x in df['Age'] if abs((x - mean) / std_dev) > threshold]

    # Visualization
    plt.figure(figsize=(10, 5))
    plt.plot(df['Age'], 'bo-', label='Data Points')
    plt.axhline(y=mean + threshold * std_dev, color='r', linestyle='--', label='Upper Threshold')
    plt.axhline(y=mean - threshold * std_dev, color='r', linestyle='--', label='Lower Threshold')
    plt.title('Z-Score Method for Outlier Detection')
    plt.xlabel('Index')
    plt.ylabel('Age Values')
    plt.legend()
    plt.show()
else:
    print("Error: Column 'Age' not found in the dataset.")

```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load the CSV file
df = pd.read_csv('Dataset_alzheimer.csv')

# Columns to analyze
columns = ['BMI', 'AlcoholConsumption', 'PhysicalActivity','DietQuality','SleepQuality','SystolicBP','DiastolicBP','CholesterolTotal','CholesterolLDL','CholesterolHDL','CholesterolTriglycerides','MMSE','FunctionalAssessment','ADL']
threshold = 3

for column in columns:
    if column in df.columns:  # Ensure column exists
        mean = np.mean(df[column])
        std_dev = np.std(df[column])

        # Detect outliers using Z-Score
        outliers_z_score = [x for x in df[column] if abs((x - mean) / std_dev) > threshold]

        # Visualization
        plt.figure(figsize=(10, 5))
        plt.plot(df[column], 'bo-', label=f'Data Points ({column})')
        plt.axhline(y=mean + threshold * std_dev, color='r', linestyle='--', label='Upper Threshold')
        plt.axhline(y=mean - threshold * std_dev, color='r', linestyle='--', label='Lower Threshold')
        plt.title(f'Z-Score Method for Outlier Detection - {column}')
        plt.xlabel('Index')
        plt.ylabel(column)
        plt.legend()
        plt.show()

        # Print detected outliers
        print(f'Outliers in {column}:', outliers_z_score)
    else:
        print(f'Column "{column}" not found in dataset.')

```
